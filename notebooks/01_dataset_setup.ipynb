```python
# 5. Report dataset statistics (counts only — computed from the manifest
# generated above, over the real dataset, not fixtures). Do not hand-edit
# these numbers; only record what this cell actually prints after running.
from collections import Counter

from evat.data.manifests import read_manifest

records = list(read_manifest("results/manifests/mvtec_ad.jsonl"))
print("total samples:", len(records))
print("by split:", Counter(r.split for r in records))
print("by category:", Counter(r.category for r in records))
print("by label:", Counter(r.label for r in records))
print("with annotation:", sum(1 for r in records if r.annotation_path))
```
